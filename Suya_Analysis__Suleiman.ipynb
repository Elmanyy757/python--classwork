### Week 5: Data Cleaning¶

### Monday: Missing Values, Duplicate Records, Data Types

### Q1.Make a copy of the data. Count the missing values in each column.

### Q2.Fill the missing Customer_Age with the median. Fill the missing Payment_Method with "Unknown".

### Q3.Count the duplicate rows, then remove them.
### Q4.Convert the Date column to a date type.


```python
# Q1. Make a copy of the data. Count the missing values in each column.
suya_orders_clean = suya_orders_data.copy()
missing_counts = suya_orders_clean.isnull().sum()
print("Missing values per column:")
print(missing_counts)
```

```python
# Q2. Fill the missing Customer_Age with the median. Fill the missing Payment_Method with "Unknown".
age_median = suya_orders_clean['Customer_Age'].median()
suya_orders_clean['Customer_Age'] = suya_orders_clean['Customer_Age'].fillna(age_median)

suya_orders_clean['Payment_Method'] = suya_orders_clean['Payment_Method'].fillna("Unknown")

# Confirm that missing values in these columns have been resolved
print(f"Median age used for filling: {age_median}")
print("Remaining missing values in Customer_Age:", suya_orders_clean['Customer_Age'].isnull().sum())
print("Remaining missing values in Payment_Method:", suya_orders_clean['Payment_Method'].isnull().sum())
```

```python
# Q3. Count the duplicate rows, then remove them.
duplicate_count = suya_orders_clean.duplicated().sum()
print(f"Number of duplicate rows found: {duplicate_count}")

suya_orders_clean = suya_orders_clean.drop_duplicates()
print(f"Shape after removing duplicates: {suya_orders_clean.shape}")
```

```python
# Q4. Convert the Date column to a date type.
suya_orders_clean['Date'] = pd.to_datetime(suya_orders_clean['Date'])
print("Data types after conversion:")
print(suya_orders_clean.dtypes[['Date']])
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set overall plot aesthetic
sns.set_theme(style="whitegrid")

# Helper for month ordering if Month column is categorical/string
month_order = [
    "January",
    "February",
    "March",
    "April",
    "May",
    "June",
    "July",
    "August",
    "September",
    "October",
    "November",
    "December",
]
```

```python
# --- Q1. Bar chart of total sales for each Branch ---
plt.figure(figsize=(10, 5))
branch_sales = (
    suya_orders_clean.groupby("Branch")["Total_NGN"].sum().reset_index()
)
sns.barplot(data=branch_sales, x="Branch", y="Total_NGN", palette="viridis")
plt.title("Total Sales (NGN) by Branch")
plt.xlabel("Branch")
plt.ylabel("Total Sales (NGN)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# --- Q2. Seaborn boxplot of Total_NGN for each Item ---
plt.figure(figsize=(10, 5))
sns.boxplot(data=suya_orders_clean, x="Item", y="Total_NGN", palette="Set2")
plt.title("Distribution of Total Sales (NGN) per Item")
plt.xlabel("Item")
plt.ylabel("Total NGN")
plt.tight_layout()
plt.show()
```

```python
# --- Q3. Line chart of total sales for each Month ---
plt.figure(figsize=(10, 5))
monthly_sales = (
    suya_orders_clean.groupby("Month")["Total_NGN"].sum().reindex(month_order).dropna()
)
plt.plot(
    monthly_sales.index,
    monthly_sales.values,
    marker="o",
    linewidth=2,
    color="b",
)
plt.title("Total Sales by Month")
plt.xlabel("Month")
plt.ylabel("Total Sales (NGN)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# --- Q4. Bar chart of the number of orders for each Item ---
plt.figure(figsize=(8, 5))
sns.countplot(
    data=suya_orders_clean,
    x="Item",
    palette="magma",
    order=suya_orders_clean["Item"].value_counts().index,
)
plt.title("Number of Orders for Each Item")
plt.xlabel("Item")
plt.ylabel("Number of Orders")
plt.tight_layout()
plt.show()
```

```python
# --- Q5. Histogram of Customer_Age ---
plt.figure(figsize=(8, 5))
sns.histplot(
    data=suya_orders_clean, x="Customer_Age", bins=20, kde=True, color="teal"
)
plt.title("Distribution of Customer Age")
plt.xlabel("Customer Age")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python
# --- Q6. Bar chart of the average Rating for each Branch ---
plt.figure(figsize=(10, 5))
branch_rating = (
    suya_orders_clean.groupby("Branch")["Rating"].mean().reset_index()
)
sns.barplot(data=branch_rating, x="Branch", y="Rating", palette="Blues_d")
plt.title("Average Rating by Branch")
plt.xlabel("Branch")
plt.ylabel("Average Rating")
plt.ylim(0, 5)  # Assuming ratings are on a scale up to 5
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# --- Q7. Seaborn countplot of Payment_Method ---
plt.figure(figsize=(8, 5))
sns.countplot(data=suya_orders_clean, x="Payment_Method", palette="pastel")
plt.title("Order Count by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

```python
# --- Q8. Scatter plot of Quantity against Total_NGN ---
plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=suya_orders_clean, x="Quantity", y="Total_NGN", alpha=0.6, color="coral"
)
plt.title("Quantity vs Total NGN")
plt.xlabel("Quantity")
plt.ylabel("Total Sales (NGN)")
plt.tight_layout()
plt.show()
```

```python
# --- Q2. Average Total_NGN for Delivery vs. Other orders ---
# In our dataset, delivery status is tracked by 'Is_Delivery' (1 for Yes, 0 for No)
q2_avg_sales = suya_orders_clean.groupby('Is_Delivery')['Total_NGN'].mean()

# Map the index to friendly names for readability
q2_avg_sales.index = q2_avg_sales.index.map({1: 'Delivery (Yes)', 0: 'Not Delivery (No)'})

print("\n--- Q2. Average Total_NGN for Delivery vs Other ---")
print(q2_avg_sales)
```

```python
# --- Q3. Payment_Method with highest average Total_NGN ---
q3_avg_pm = (
    suya_orders_clean.groupby("Payment_Method")["Total_NGN"]
    .mean()
    .sort_values(ascending=False)
)

print("\n--- Q3. Average Total_NGN by Payment Method ---")
print(q3_avg_pm)
print(
    f"\nHighest Payment Method: {q3_avg_pm.index[0]} "
    f"(Avg NGN {q3_avg_pm.iloc[0]:,.2f})"
)
```

```python
import pandas as pd

if 'suya_orders_clean' not in globals():
    # Load the original data
    suya_orders_data = pd.read_csv('/content/drive/MyDrive/suya_orders.csv')
    suya_orders_clean = suya_orders_data.copy()

    # Re-apply key cleaning steps
    age_median = suya_orders_clean['Customer_Age'].median()
    suya_orders_clean['Customer_Age'] = suya_orders_clean['Customer_Age'].fillna(age_median)
    suya_orders_clean['Payment_Method'] = suya_orders_clean['Payment_Method'].fillna("Unknown")
    suya_orders_clean = suya_orders_clean.drop_duplicates()
    suya_orders_clean['Date'] = pd.to_datetime(suya_orders_clean['Date'])
    suya_orders_clean['Is_Delivery'] = suya_orders_clean['Is_Delivery'].map({'Yes': 1, 'No': 0})

# 1. Mean, Median, and Mode of Total_NGN
mean_total = suya_orders_clean['Total_NGN'].mean()
median_total = suya_orders_clean['Total_NGN'].median()
mode_total = suya_orders_clean['Total_NGN'].mode()[0]

print(f"Total_NGN Mean: {mean_total:.2f} NGN")
print(f"Total_NGN Median: {median_total:.2f} NGN")
print(f"Total_NGN Mode: {mode_total} NGN")
```

```python
# 2. Standard Deviation of Customer_Age
std_age = suya_orders_clean['Customer_Age'].std()

print(f"Customer_Age Standard Deviation: {std_age:.2f} years")
```

```python
# 3. Probability that a random order is a delivery
# Works whether Is_Delivery is numeric (1 / 0) or categorical ('Yes' / 'No')
if suya_orders_clean['Is_Delivery'].dtype in ['int64', 'float64']:
    prob_delivery = (suya_orders_clean['Is_Delivery'] == 1).mean()
else:
    prob_delivery = (suya_orders_clean['Is_Delivery'] == 'Yes').mean()

print(f"Probability of Delivery: {prob_delivery:.4f} ({prob_delivery * 100:.2f}%)")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# -------------------------------------------------------------------
# 1. Compare Average Total_NGN of a 100-Order Sample vs Population
# -------------------------------------------------------------------
sample_100 = suya_orders_clean['Total_NGN'].sample(n=100, random_state=42)

pop_mean = suya_orders_clean['Total_NGN'].mean()
sample_100_mean = sample_100.mean()

print(f"Population Mean Total_NGN (All Orders): {pop_mean:.2f} NGN")
print(f"Sample Mean Total_NGN (100 Orders):     {sample_100_mean:.2f} NGN")
print(f"Difference:                             {sample_100_mean - pop_mean:.2f} NGN\n")


# -------------------------------------------------------------------
# 2. Draw a Histogram of Total_NGN
# -------------------------------------------------------------------
plt.figure(figsize=(10, 5))
sns.histplot(suya_orders_clean['Total_NGN'], bins=30, kde=True, color='skyblue', edgecolor='black')
plt.title('Histogram of Total_NGN Distribution')
plt.xlabel('Total Order Amount (NGN)')
plt.ylabel('Frequency')
plt.grid(axis='y', alpha=0.3)
plt.show()


# -------------------------------------------------------------------
# 3. 95% Confidence Interval for Average Total_NGN (200 Orders Sample)
# Formula: mean ± 1.96 * (std / sqrt(200))
# -------------------------------------------------------------------
sample_200 = suya_orders_clean['Total_NGN'].sample(n=200, random_state=42)

sample_200_mean = sample_200.mean()
sample_200_std = sample_200.std()

margin_of_error = 1.96 * (sample_200_std / np.sqrt(200))
lower_bound = sample_200_mean - margin_of_error
upper_bound = sample_200_mean + margin_of_error

print("--- 95% Confidence Interval (n=200) ---")
print(f"Sample Mean (n=200):     {sample_200_mean:.2f} NGN")
print(f"Sample Std Deviation:    {sample_200_std:.2f} NGN")
print(f"Margin of Error:         ±{margin_of_error:.2f} NGN")
print(f"95% Confidence Interval: ({lower_bound:.2f} NGN, {upper_bound:.2f} NGN)")
```
